```python
import numpy as np

# Numpy Array Properties
```

```python
arr = np.array([1,2,3,4.0,5])

# shape : check dimension(row x cols)
print(arr.shape)

# size: to print total elements
print(arr.size)

# dtype : print datatype
print(arr.dtype)

# ndim : to print dimension
print(arr.ndim ,"D Array")

# astype: alternate to dtype to convert explicitly data type
intArr = arr.astype(np.int64)
print(intArr)
```
