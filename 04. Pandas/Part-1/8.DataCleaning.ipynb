```python
import pandas as pd
```

```python
data = pd.read_csv("raw_data.csv")

data
```

```python
# identify missing value : isnull() or isna()
data.isnull() # TRUE value are missing
```

```python
# count total missing value in all table
data.isna().sum()
```

```python
# removing from tables
data.dropna() # display row without missing value
```

```python
#  fill null value with other value
avgAge = data["age"].mean()
data.fillna(avgAge)
```

```python
# forward and backward filling

data.ffill() # upper copy to lower
data.bfill() # lower copy to upper
```

```python
# ################ Handling Duplicates #################################
# duplicated()

data.duplicated()
data["country"].duplicated()

# drop_duplicated()
orgcopy = data.copy() # contain duplicate

data.drop_duplicates() # chnage in copy
data.drop_duplicates(inplace=True) # chnage in real data

data # see original get inplaced.
```
