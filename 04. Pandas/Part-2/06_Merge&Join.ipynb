```python
import pandas as pd
```

```python
customer = pd.DataFrame({
    "cid" : [1,2,3] ,
    "name" : ["ALpha" , "Beta " , "Gamma"]
})


order = pd.DataFrame({
    "oid" : [101,102,103,104] ,
    "cid" : [2,1,4,5] ,
    "amount" : [256, 341, 678, 890]
})

# merge is simliar to join in SQL use to join two or more tables 
# 4types: inner, outer, left , right

pd.merge(customer, order , on="cid" , how="left") # keep left all value
pd.merge(customer, order , on="cid" , how="inner") # common values
```
