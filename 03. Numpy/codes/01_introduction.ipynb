```python
import numpy as np
import time

# create ndarray of numpy

arr = np.array([1,2,3,4,5])
print(arr , type(arr))
```

```python
# calculate performance (numpy array vs list)
size = 1_000_000 
# python list
pylist = list(range(size))

# calculate Square time 
startTimeList = time.time()
sqr = [x**2 for x in pylist]
endTimeList = time.time()
print(f"Time Taken by List is {endTimeList - startTimeList} seconds")


# numpy array
ndarr = np.array(pylist)
# calculate time 
startTimeList = time.time()
sqr = ndarr ** 2 #vectorization of Numpy
endTimeList = time.time()

print(f"Time Taken by Numpy Array is {endTimeList - startTimeList} seconds")
```
