```python
import pandas as pd
```

```python
data = pd.read_csv("globalAirQuality.csv")

# selection
data.head(10) # top 10 rows print


# --------------------------- COLUMNS BASED ACCESS --------------------------------
# specific column based access 
data[["city","aqi"]]
```

```python
# ----------------------------------- ROW SELECTION --------------------------------
# loc (labelled based access) vs iloc (index based access)
data.loc[0]
data.loc[0:5] # start: END inclusive

data.iloc[0]
data.iloc[0:5] # end exclusive
```

```python
# selecting cells using iloc vs loc

data.loc[0:5 , "city"] # range , column label name
data.loc[0:5 , ["country" , "city" , "aqi"]]


# using iloc
# data.iloc[0:5 , "city"] ------------> ERROR

# data.columns ----------> to print column index
data.columns
data.iloc[0:5 , [1,2,11]] # range of rows , column index 
```

```python
#  access cell using at and iat

# using at (for labelled)
data.at[0,"city"]

# using iat (for index)
data.iat[0,1]


# NOTE: ALWAYS KEEP COPY OF ORG DATA BEFORE CHANGE IT> >copy() FUNCTION
```

```python
# filtering DATA 

# use list of list
data[ data["aqi"] > 100]

data[ (data["aqi"] > 100) & (data["temperature"] > 35)]

# display desired column
data[(data["aqi"] > 100) & (data["temperature"] > 35)][["city","country"]]
```

```python
# QUERY FILTER (best method)

data.query("aqi > 150 & temperature > 35")[["city" , "country"]]

aqi_good = 30
# print best city to live
data.query("aqi < @aqi_good")[["city" , "country" , "aqi"]]
```
