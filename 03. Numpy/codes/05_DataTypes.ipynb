```python
import numpy as np
```

```python
# Data Types

# 1. int32 or int64
arr_int64 = np.array([1, 2, 3, 4, 5])
print(arr_int64.dtype)

# 2. float64
arr_float64 = np.array([1,2,3,4.7,5])
print(arr_float64.dtype)

```

```python
# 3. complex datatype

arr1 = np.array([3 + 2j])
arr2 = np.array([5 + 2j])

print(arr1+arr2 , arr1.dtype)
```

```python
# 4. String(S and U) and Object
obj = np.array([{1,2,3 }, 3.14 , "STR", 345] )
print(obj.dtype)


strg = np.array(['hello' , 'hi'])
print(strg.dtype)
```
