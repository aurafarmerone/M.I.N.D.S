```python
import pandas as pd
```

```python
# Series ( 1D Labelled Data)

# index based access
s = pd.Series([87,23,56,13,67])
print("Using idx :" , s[2])

# labelled base access
s = pd.Series([23,19,45,19,21] , index=["Alpha" , "Beta" , "Delta" , "Sigma" , "Omega"])
print("Using Label :", s["Omega"])
```

```python
# Series Properties:

# 1. Homogeneous (all value converted into one datatypes)

s = pd.Series([23,19,45,19,21 , "XYZ"] , index=["Alpha" , "Beta" , "Delta" , "Sigma" , "Omega" , 99])
print(s , s.dtype)
```

```python
# 2. vectorized operation

s1 = pd.Series([23,19,45,19,21])
s2 = pd.Series([43,79,435,99,11])

print(s1+s2) #vectorized operation supported 
```

```python
# 3. Handle missing value with NaN(see later)

# 4. mutuable value yet immutable size
s1 = pd.Series([23,19,45,19,21])

s1[0] = 999 # mutuable value
print(s1)


s1.drop(0) # immutuable size even after drop
print(s1)

changed = s1.drop(0) # drop here when new Series created
print(changed)
```
