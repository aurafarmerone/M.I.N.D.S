```python
import pandas as pd
```

```python
# ways to create data in data frames
# 1. using dictionary
info = {
    "Name" : ["a", "b", "c"] ,
    "CGPA" : [9.9 , 9.7 , 9.0]
}

data = pd.DataFrame(info)
print(data)

# 2. using list of lists

data = pd.DataFrame([ ["A" , 9.9] , ["B" , 9.7] , ["C" , 8.7]] , columns=["Name" , "CGPA"])

print(data)
```
