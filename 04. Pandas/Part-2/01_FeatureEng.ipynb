```python
data = pd.read_csv("employee_data.csv")
```

```python
d = data.copy()

# def logic():
#     if d["Salary"] < 50000:
#         return "20%"
#     else:
#         return "10%"
    
# apply
d["tax"] = d["Salary"].apply(lambda x : "20%" if x > 50000 else "10%")
d



# map

dep_map = {"HR":"Faltu" , "IT" : "ITI wala"}
d["Department"] = d["Department"].map(dep_map)
d

#  replace function


d["Department"] = d["Department"].replace("ITI wala" , "Nhi pata")
d
```
