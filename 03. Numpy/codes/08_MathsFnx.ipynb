```python
# ================================
# NumPy – Important Mathematical Functions
# ================================

import numpy as np

# Sample arrays
a = np.array([1, 2, 3, 4, 5])
b = np.array([1.2, 2.7, 3.5, 4.9])
c = np.array([1, 2, 2, 3, 3, 3])

print("Array a:", a)
print("Array b:", b)
print("Array c:", c)

# ----------------
# Aggregation
# ----------------
print("\n--- Aggregation ---")
print("sum:", np.sum(a))
print("prod:", np.prod(a))
print("min:", np.min(a))
print("max:", np.max(a))
print("mean:", np.mean(a))
print("std:", np.std(a))
print("var:", np.var(a))
print("median:", np.median(a))
print("argmin:", np.argmin(a))
print("argmax:", np.argmax(a))

# ----------------
# Power
# ----------------
print("\n--- Power ---")
print("square:", np.square(a))
print("sqrt:", np.sqrt(a))
print("power(a, 2):", np.power(a, 2))

# ----------------
# Rounding
# ----------------
print("\n--- Rounding ---")
print("round:", np.round(b))
print("floor:", np.floor(b))
print("ceil:", np.ceil(b))
print("trunc:", np.trunc(b))

# ----------------
# Log & Exponential
# ----------------
print("\n--- Log & Exponential ---")
print("log:", np.log(a))
print("log10:", np.log10(a))
print("log2:", np.log2(a))
print("exp:", np.exp(a))

# ----------------
# Extras
# ----------------
print("\n--- Extras ---")
print("unique:", np.unique(c))
print("sort:", np.sort(b))
print("abs:", np.abs([-3, -1, 2]))
```
